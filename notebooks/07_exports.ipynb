# Export generated charts for a simulator

`ChartExporter` (`src/donderformer/exporter.py`) turns the charts generated by one or more runs into song folders: one copy of the audio plus one `.tja` per source, so the same song can be played as charted by each model (and by the human) side by side.

```text
outputs/exports/<datetime>-<alias>/
├── export.json
└── Idol/
    ├── Idol.ogg
    ├── Idol (lstm).tja
    └── Idol (human).tja
```

Upload the export folder to the simulator. Every export is a new folder, so nothing is overwritten. Needs the raw audio under `data/raw/` (or `DONDER_DATA`).

````python
from donderformer.exporter import ChartExporter
from donderformer.paths import DataPaths, outputs_dir

paths = DataPaths()
runs = outputs_dir() / "runs" / "taikonation"
latest = sorted(p for p in runs.iterdir() if (p / "eval" / "ar_test" / "charts").is_dir())[-1]

exporter = ChartExporter(
    alias="lstm",                                   # names the export folder
    sources={                                       # label -> folder of generated <chart_id>.npz
        "lstm": latest / "eval" / "ar_test" / "charts",
        # "transformer": outputs_dir() / "runs" / "transformer" / "<run>" / "eval" / "ar_test" / "charts",
    },
    paths=paths,
    split="oni_v1",                                 # the split the runs generated from
    version="clean",                                # "clean" (back-to-back notes removed) or "raw"
    include_human=True,                             # add the human chart for comparison
    songs=None,                                     # exact titles, e.g. ["Idol", "YES or YES"]; None = all
    n_songs=None,                                   # first N songs (manifest order); ignored when songs is set
)
print(f"{len(exporter.rows)} songs selected from", {label: str(f) for label, f in exporter.sources.items()})
````

````python
export_dir = exporter.export()
````
